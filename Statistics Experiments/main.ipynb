```python
import pandas as pd
import numpy as np
import statistics as st
import matplotlib.pyplot as plt
import random
```

```python
df = pd.read_excel("Class_activity_sheet.xlsx")
df
```

Group number = 7

We will deal with Length of movies

### Extracting Population data

```python
Population = df['Length of fav movie (in mins)']
gr_no = 7
```

# Step 1

### Getting the mean and std of population data

```python
mean = Population.mean()
std = st.pstdev(Population)
print(f"The mean of the population = {mean}\nThe standard deviation of the {std}")
```

### Drawing a histogram

```python
plt.hist(Population,bins=30)
plt.show()
```

# Step 2

```python
n = 5*gr_no
```

```python
samples = np.zeros([100,n])
for i in range(100):
    samples[i] = np.random.choice(Population,size= n,replace =False)

samples
```

```python
means = np.mean(samples,axis=1)
means
```

```python
plt.hist(means,bins =40,color='maroon')
plt.title("Histogram of Sample Means")
plt.xlabel("Sample Means (in mins)")
plt.ylabel("Frequencies")
plt.grid()
plt.show()
```

```python
avg_mean = means.mean()
avg_mean
```

```python
stds = np.std(samples,axis=1)
stds
```

```python
ste = std/np.sqrt(n)
print(f'The standard error is {ste}')
```
